# Customer Clustering

```python
data.info()
```

```python
data_customers = data.groupby("Customer ID")[["Sales","Quantity","Discount","Order ID","Profit"]]\
    .agg({'Profit':'sum','Sales':'sum','Order ID':'count','Quantity':'sum','Discount':'mean'})\
    .reset_index()
```

```python
data_customers
```

```python
features = data_customers[['Sales', 'Quantity', 'Discount', 'Order ID','Profit']]

scaler = StandardScaler()
scaled_features = scaler.fit_transform(features)
```

```python
inertia = []
K = range(2,11)
for k in K:
    kmeans = KMeans(n_clusters = k,random_state=7)
    kmeans.fit(scaled_features)
    inertia.append(kmeans.inertia_)
    
plt.figure(figsize=(17,9))    

plt.subplot(1,2,1)
sns.lineplot(x=K,y=inertia)
plt.title("Elbow method for optimal k")
plt.xlabel("K")
plt.ylabel("Inertia")

plt.subplot(1,2,2)
model = KMeans(random_state=7)
elbow_visualizer = KElbowVisualizer(model, k=(1, 11), timings=False)
elbow_visualizer.fit(scaled_features)
elbow_visualizer.show()
```

```python
silhouette_scores = []
for k in K:
    kmeans= KMeans(n_clusters=k, random_state=7)
    kmeans.fit(scaled_features)
    labels = kmeans.labels_
    silhouette_scores.append(silhouette_score(scaled_features, labels))
    
plt.figure(figsize=(17,9))
plt.plot(K, silhouette_scores,"go-")
plt.title("silhouette method")
plt.xlabel("number of clusters")
plt.ylabel("silhouette score")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
kmeans = KMeans(n_clusters=4, random_state=7) 
kmeans.fit(scaled_features)
data_customers['Cluster'] = kmeans.labels_
```

## Evaluating Clusters

```python

pca = PCA(n_components=2)
X_pca = pca.fit_transform(scaled_features)

plt.figure(figsize=(8, 6))
cmap = plt.get_cmap('viridis', len(np.unique(data_customers['Cluster'])))
scatter = plt.scatter(X_pca[:, 0], X_pca[:, 1], c=data_customers['Cluster'], cmap=cmap, s=50)
plt.title("Product Clusters (PCA Projection)")
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.colorbar(scatter,ticks=np.unique(data_customers['Cluster']), label='Cluster')
plt.grid(True)
plt.show()
```

```python
score = silhouette_score(scaled_features, kmeans.labels_)
print(f'Silhouette Score for {kmeans.n_clusters} clusters: {score}')
```

```python
silhouette_visualizer = SilhouetteVisualizer(kmeans, colors='yellowbrick')
silhouette_visualizer.fit(scaled_features)
silhouette_visualizer.show()
```

```python
cluster_summary = data_customers.groupby('Cluster')[['Sales', 'Quantity', 'Discount', 'Order ID','Profit']].mean()
print(cluster_summary)
```

```python
data_customers["Cluster"].value_counts()
```

### **Cluster Insights:-**
    - The clusters divide customers into 4 groups, the most important of them is group 2 which the group with high sales and high profit and could be called the high paying customers and should be priority for retention and marketing.

    - group 3 has the high average discount and also a negative profit so these could be called discount chasers.

    - group 1 has moderate average discount and low profit even with moderate sales, so could also be called discount chasers

    - group 0 are the low performing customers with not so high of a profit or sales

```python
cluster_labels = {
    0: "low performing customers",
    1: "Semi discount chasers",
    2: "High paying customers",
    3: "Discount Chasers"
}

data_customers['Cluster Label'] = data_customers['Cluster'].map(cluster_labels)
data_customers.to_csv(os.path.join(data_path, "data_customers.csv"), index=False)
```

# Products clustering

```python
data_products = data.groupby("Product Name")[["Sales","Profit","Quantity","Discount"]]\
    .agg({'Sales':'sum','Profit':'sum','Quantity':'sum','Discount':'mean'})\
    .reset_index()
```

```python
features = data_products.loc[:,["Discount","Sales","Profit","Quantity"]]

scaler = StandardScaler()
scaled_features = scaler.fit_transform(features)
```

```python
inertia = []
K = range(2,11)
for k in K:
    kmeans = KMeans(n_clusters = k,random_state=7)
    kmeans.fit(scaled_features)
    inertia.append(kmeans.inertia_)
    
plt.figure(figsize=(17,9))    

plt.subplot(1,2,1)
sns.lineplot(x=K,y=inertia)
plt.title("Elbow method for optimal k")
plt.xlabel("K")
plt.ylabel("Inertia")

plt.subplot(1,2,2)
model = KMeans(random_state=7)
elbow_visualizer = KElbowVisualizer(model, k=(1, 11), timings=False)
elbow_visualizer.fit(scaled_features)
elbow_visualizer.show()
```

```python
silhouette_scores = []
for k in K:
    kmeans= KMeans(n_clusters=k, random_state=7)
    kmeans.fit(scaled_features)
    labels = kmeans.labels_
    silhouette_scores.append(silhouette_score(scaled_features, labels))
    
plt.figure(figsize=(17,9))
plt.plot(K, silhouette_scores,"go-")
plt.title("silhouette method")
plt.xlabel("number of clusters")
plt.ylabel("silhouette score")
plt.grid(True)
plt.tight_layout()
plt.show()
```

#### Both methods showed 4 to be optimal but i found 3 to have better interpretation

```python
kmeans_final = KMeans(n_clusters= 3,random_state=7)
kmeans_final.fit(scaled_features)
```

```python
data_products['Cluster'] = kmeans_final.labels_
```

## Evaluating Clusters
